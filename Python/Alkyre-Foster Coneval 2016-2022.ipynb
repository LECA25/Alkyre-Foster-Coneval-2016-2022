```python
## Especie ##
especie16 = ingresonomonetariodeflac16
especie18 = ingresonomonetariodeflac18
especie20 = ingresonomonetariodeflac20
especie22 = ingresonomonetariodeflac22

# Mantener observaciones en especie
especie16 = especie16.loc[especie16["especie"] == 1]
especie18 = especie18.loc[especie18["especie"] == 1]
especie20 = especie20.loc[especie20["especie"] == 1]
especie22 = especie22.loc[especie22["especie"] == 1]

# Agrupar por identificador del hogar
# 2016
gnm_cols16 = [col for col in especie16.columns if col.startswith('gnm_')]
especie16 = especie16.groupby("idhogar")[gnm_cols16].sum().reset_index()
# 2018
gnm_cols18 = [col for col in especie18.columns if col.startswith("gnm_")]
especie18 = especie18.groupby("idhogar")[gnm_cols18].sum().reset_index()
# 2020
gnm_cols20 = [col for col in especie20.columns if col.startswith("gnm_")]
especie20 = especie20.groupby("idhogar")[gnm_cols20].sum().reset_index()
# 2022
gnm_cols22 = [col for col in especie22.columns if col.startswith("gnm_")]
especie22 = especie22.groupby("idhogar")[gnm_cols22].sum().reset_index()

# Renombrar variables
nuevos_nombres_especie = {"gnm_alcoholytabaco":"gnm_alcoholytabacoespecie","gnm_alimentos":"gnm_alimentosespecie",
                                      'gnm_vestidoycalzado':'gnm_vestidoycalzadoespecie','gnm_viviendas':'gnm_viviendasespecie',
                                      'gnm_alimpieza':'gnm_alimpiezaespecie','gnm_cristaleria':'gnm_cristaleriaespecie',
                                      'gnm_edomesticos':'gnm_edomesticosespecie','gnm_salud':'gnm_saludespecie',
                                      'gnm_transportepublico':'gnm_transportepublicoespecie','gnm_transporteforaneo':'gnm_transporteforaneoespecie',
                                      'gnm_comunicaciones':'gnm_comunicacionesespecie','gnm_educacionyrecreacion':'gnm_educacionyrecreacionespecie',
                                      'gnm_educacionbasica':'gnm_educacionbasicaespecie','gnm_cuidadopersonal':'gnm_cuidadopersonalespecie',
                                      'gnm_accesoriospersonales':'gnm_accesoriospersonalesespecie','gnm_otrosgastos':'gnm_otrosgastosespecie',
                                      'gnm_regalosotorgados':'gnm_regalosotorgadosespecie'}

lista_df_especie = [especie16,especie18,especie20,especie22]
for df in lista_df_especie:
    df.rename(columns=nuevos_nombres_especie,inplace=True)

## Regalos ##
regalos16 = ingresonomonetariodeflac16
regalos18 = ingresonomonetariodeflac18
regalos20 = ingresonomonetariodeflac20
regalos22 = ingresonomonetariodeflac22

# Mantener observaciones de regalo
regalos16 = regalos16.loc[regalos16["regalos"] == 1]
regalos18 = regalos18.loc[regalos18["regalos"] == 1]
regalos20 = regalos20.loc[regalos20["regalos"] == 1]
regalos22 = regalos22.loc[regalos22["regalos"] == 1]

# Agrupar por identificador del hogar
# 2016
gnm_colreg16 = [col for col in regalos16.columns if col.startswith("gnm_")]
regalos16 = regalos16.groupby("idhogar")[gnm_colreg16].sum().reset_index()
# 2018
gnm_colreg18 = [col for col in regalos18.columns if col.startswith("gnm_")]
regalos18 = regalos18.groupby("idhogar")[gnm_colreg18].sum().reset_index()
# 2020
gnm_colreg20 = [col for col in regalos20.columns if col.startswith("gnm_")]
regalos20 = regalos20.groupby("idhogar")[gnm_colreg20].sum().reset_index()
# 2022
gnm_colreg22 = [col for col in regalos22.columns if col.startswith("gnm_")]
regalos22 = regalos22.groupby("idhogar")[gnm_colreg22].sum().reset_index()

# Renombrar variables
nuevos_nombres_regalos = {"gnm_alcoholytabaco":"gnm_alcoholytabacoregalos","gnm_alimentos":"gnm_alimentosregalos",
                                      'gnm_vestidoycalzado':'gnm_vestidoycalzadoregalos','gnm_viviendas':'gnm_viviendasregalos',
                                      'gnm_alimpieza':'gnm_alimpiezaregalos','gnm_cristaleria':'gnm_cristaleriaregalos',
                                      'gnm_edomesticos':'gnm_edomesticosregalos','gnm_salud':'gnm_saludregalos',
                                      'gnm_transportepublico':'gnm_transportepublicoregalos','gnm_transporteforaneo':'gnm_transporteforaneoregalos',
                                      'gnm_comunicaciones':'gnm_comunicacionesregalos','gnm_educacionyrecreacion':'gnm_educacionyrecreacionregalos',
                                      'gnm_educacionbasica':'gnm_educacionbasicaregalos','gnm_cuidadopersonal':'gnm_cuidadopersonalregalos',
                                      'gnm_accesoriospersonales':'gnm_accesoriospersonalesregalos','gnm_otrosgastos':'gnm_otrosgastosregalos',
                                      'gnm_regalosotorgados':'gnm_regalosotorgadosregalos'}

lista_regalos = [regalos16,regalos18,regalos20,regalos22]

for df in lista_regalos:
    df.rename(columns=nuevos_nombres_regalos,inplace=True)

# Ingreso Corriente Total
# Concentrado del Hogar
conc_hog16 = concentradohogar16
conc_hog18 = concentradohogar18
conc_hog20 = concentradohogar20
conc_hog22 = concentradohogar22

# Lista del ingreso corriente total
lista_ingresocorrientetotal = [conc_hog16,conc_hog18,conc_hog20,conc_hog22]
# Bucle
for i, df in enumerate(lista_ingresocorrientetotal):
    # Indentificador del Hogar
    df["idhogar"] = df["folioviv"].astype(str) + df["foliohog"].astype(str)
    lista_ingresocorrientetotal[i] = df

# Mantener variables 
mant_var_concentrado_hogar = ["idhogar","tam_loc","factor","tot_integ","est_dis","upm","ubica_geo"]
conc_hog16 = conc_hog16[mant_var_concentrado_hogar]
conc_hog18 = conc_hog18[mant_var_concentrado_hogar]
conc_hog20 = conc_hog20[mant_var_concentrado_hogar]
conc_hog22 = conc_hog22[mant_var_concentrado_hogar]

# Unir Bases 
# 2016
ingresocorrientetotal16 = pd.merge(left=conc_hog16,right=ingresodeflactado16,on="idhogar",how="outer")
ingresocorrientetotal16 = pd.merge(left=ingresocorrientetotal16,right=especie16,on="idhogar",how="left")
ingresocorrientetotal16 = pd.merge(left=ingresocorrientetotal16,right=regalos16,on="idhogar",how="left")
# 2018
ingresocorrientetotal18 = pd.merge(left=conc_hog18,right=ingresodeflactado18,on="idhogar",how="outer")
ingresocorrientetotal18 = pd.merge(left=ingresocorrientetotal18,right=especie18,on="idhogar",how="left")
ingresocorrientetotal18 = pd.merge(left=ingresocorrientetotal18,right=regalos18,on="idhogar",how="left")
# 2020
ingresocorrientetotal20 = pd.merge(left=conc_hog20,right=ingresodeflactado20,on="idhogar",how="outer")
ingresocorrientetotal20 = pd.merge(left=ingresocorrientetotal20,right=especie20,on="idhogar",how="left")
ingresocorrientetotal20 = pd.merge(left=ingresocorrientetotal20,right=regalos20,on="idhogar",how="left")
# 2022
ingresocorrientetotal22 = pd.merge(left=conc_hog22,right=ingresodeflactado22,on="idhogar",how="outer")
ingresocorrientetotal22 = pd.merge(left=ingresocorrientetotal22,right=especie22,on="idhogar",how="left")
ingresocorrientetotal22 = pd.merge(left=ingresocorrientetotal22,right=regalos22,on="idhogar",how="left")

# Pagos en especie
pagoenregalos = ["gnm_alcoholytabacoregalos","gnm_alimentosregalos",'gnm_vestidoycalzadoregalos','gnm_viviendasregalos',
                'gnm_alimpiezaregalos','gnm_cristaleriaregalos','gnm_edomesticosregalos','gnm_saludregalos',
                'gnm_transportepublicoregalos','gnm_transporteforaneoregalos','gnm_comunicacionesregalos','gnm_educacionyrecreacionregalos',
                'gnm_educacionbasicaregalos','gnm_cuidadopersonalregalos','gnm_accesoriospersonalesregalos','gnm_otrosgastosregalos',
                'gnm_regalosotorgadosregalos']

pagoenespecie = ["gnm_alcoholytabacoespecie","gnm_alimentosespecie",'gnm_vestidoycalzadoespecie','gnm_viviendasespecie',
                'gnm_alimpiezaespecie','gnm_cristaleriaespecie','gnm_edomesticosespecie','gnm_saludespecie',
                'gnm_transportepublicoespecie','gnm_transporteforaneoespecie','gnm_comunicacionesespecie','gnm_educacionyrecreacionespecie',
                'gnm_educacionbasicaespecie','gnm_cuidadopersonalespecie','gnm_accesoriospersonalesespecie','gnm_otrosgastosespecie',
                'gnm_regalosotorgadosespecie']

# Actualizar lista 
lista_ingresocorrientetotal = [ingresocorrientetotal16,ingresocorrientetotal18,ingresocorrientetotal20,ingresocorrientetotal22]
# Bucle
for i, df in enumerate(lista_ingresocorrientetotal):
    # Vive en zona rural
    df["rural"]  = df["tam_loc"].apply(lambda x: 0 if x <= 3 else(1 if x == 4 else np.nan))
    # Pago en especie deflactado
    df["pago_especie"] = df[pagoenespecie].sum(axis=1,skipna=True)
    # Pago en regalos deflactado
    df["pago_regalos"] = df[pagoenregalos].sum(axis=1,skipna=True)
    # Ingreso no monetario deflactado
    df["inomonetario"] = df[["pago_especie","pago_regalos"]].sum(axis=1)
    df["icorrientetotal"] = df[["inomonetario","ingresomonetariop"]].sum(axis=1)
    lista_ingresocorrientetotal[i] = df

# Tamaño del hogar
tamañohogar16 = poblacion16
tamañohogar18 = poblacion18
tamañohogar20 = poblacion20
tamañohogar22 = poblacion22

# Lista tamaño del hogar
lista_tamaño_hogar = [tamañohogar16,tamañohogar18,tamañohogar20,tamañohogar22]
# bucle
for i, df in enumerate(lista_tamaño_hogar):
    df["idhogar"] = df["folioviv"].astype(str) + df["foliohog"].astype(str)
    lista_tamaño_hogar[i] = df

# Filtrar por parentesco
tamañohogar16 = tamañohogar16.loc[~((tamañohogar16["parentesco"].between(400,499)) | (tamañohogar16["parentesco"].between(700,799)))].reset_index(drop=True)
tamañohogar18 = tamañohogar18.loc[~((tamañohogar18["parentesco"].between(400,499)) | (tamañohogar18["parentesco"].between(700,799)))].reset_index(drop=True)
tamañohogar20 = tamañohogar20.loc[~((tamañohogar20["parentesco"].between(400,499)) | (tamañohogar20["parentesco"].between(700,799)))].reset_index(drop=True)
tamañohogar22 = tamañohogar22.loc[~((tamañohogar22["parentesco"].between(400,499)) | (tamañohogar22["parentesco"].between(700,799)))].reset_index(drop=True)

# Actualizar lista
lista_tamaño_hogar = [tamañohogar16,tamañohogar18,tamañohogar20,tamañohogar22]
# Bucle
for i, df in enumerate(lista_tamaño_hogar):
    df["integrante"] = 1
    df["tot_integrantes"] = df.groupby("idhogar")["integrante"].transform("sum")
    lista_tamaño_hogar[i] = df

# Agrupar 
tamañohogar16 = tamañohogar16.groupby("idhogar")["tot_integrantes"].max()
tamañohogar18 = tamañohogar18.groupby("idhogar")["tot_integrantes"].max()
tamañohogar20 = tamañohogar20.groupby("idhogar")["tot_integrantes"].max()
tamañohogar22 = tamañohogar22.groupby("idhogar")["tot_integrantes"].max()

# Bienestar Económico
lp_ingresos16 = pd.merge(left=ingresocorrientetotal16,right=tamañohogar16,on="idhogar",how="left")
lp_ingresos18 = pd.merge(left=ingresocorrientetotal18,right=tamañohogar18,on="idhogar",how="left")
lp_ingresos20 = pd.merge(left=ingresocorrientetotal20,right=tamañohogar20,on="idhogar",how="left")
lp_ingresos22 = pd.merge(left=ingresocorrientetotal22,right=tamañohogar22,on="idhogar",how="left")

Lineas_pobreza = {
    (2016,"Rural") : {"Linea de Pobreza": 2019.01},
    (2016,"Urbana") : {"Linea de Pobreza": 2893.10},
    (2018,"Rural") : {"Linea de Pobreza": 2288.44},
    (2018,"Urbana") : {"Linea de Pobreza": 3287.59},
    (2020,"Rural") : {"Linea de Pobreza": 2503.19},
    (2020,"Urbana") : {"Linea de Pobreza": 3536.85},
    (2022,"Rural") : {"Linea de Pobreza": 2928.47},
    (2022,"Urbana") : {"Linea de Pobreza": 4105.11},
}

# Lista lp_ingresos
lista_lp_ingresos = [lp_ingresos16,lp_ingresos18,lp_ingresos20,lp_ingresos22]
# Bucle
for i, df in enumerate(lista_lp_ingresos):
    df["ict_percapita"] = df["icorrientetotal"]/df["tot_integrantes"]
    df["pobres_lpi"] = np.nan
    lista_lp_ingresos[i] = df

# Linea de Pobreza por Ingresos
# 2016
lp_ingresos16.loc[((lp_ingresos16["ict_percapita"] < Lineas_pobreza[(2016,"Rural")]["Linea de Pobreza"]) & (lp_ingresos16["rural"] == 1)) |
                  ((lp_ingresos16["ict_percapita"] < Lineas_pobreza[(2016,"Urbana")]["Linea de Pobreza"]) & (lp_ingresos16["rural"] == 0)), "pobres_lpi"] = 1
lp_ingresos16.loc[(((lp_ingresos16["ict_percapita"] >= Lineas_pobreza[(2016,"Rural")]["Linea de Pobreza"]) & (lp_ingresos16["rural"] == 1)) & ~(lp_ingresos16["ict_percapita"].isna())) |
                  (((lp_ingresos16["ict_percapita"] >= Lineas_pobreza[(2016,"Urbana")]["Linea de Pobreza"]) & (lp_ingresos16["rural"] == 0)) & ~(lp_ingresos16["ict_percapita"].isna())), "pobres_lpi"] = 0
# 2018
lp_ingresos18.loc[((lp_ingresos18["ict_percapita"] < Lineas_pobreza[(2018,"Rural")]["Linea de Pobreza"]) & (lp_ingresos18["rural"] == 1)) |
                  ((lp_ingresos18["ict_percapita"] < Lineas_pobreza[(2018,"Urbana")]["Linea de Pobreza"]) & (lp_ingresos18["rural"] == 0)), "pobres_lpi"] = 1
lp_ingresos18.loc[(((lp_ingresos18["ict_percapita"] >= Lineas_pobreza[(2018,"Rural")]["Linea de Pobreza"]) & (lp_ingresos18["rural"] == 1)) & ~(lp_ingresos18["ict_percapita"].isna())) |
                  (((lp_ingresos18["ict_percapita"] >= Lineas_pobreza[(2018,"Urbana")]["Linea de Pobreza"]) & (lp_ingresos18["rural"] == 0)) & ~(lp_ingresos18["ict_percapita"].isna())), "pobres_lpi"] = 0
# 2020
lp_ingresos20.loc[((lp_ingresos20["ict_percapita"] < Lineas_pobreza[(2020,"Rural")]["Linea de Pobreza"]) & (lp_ingresos20["rural"] == 1)) |
                  ((lp_ingresos20["ict_percapita"] < Lineas_pobreza[(2020,"Urbana")]["Linea de Pobreza"]) & (lp_ingresos20["rural"] == 0)), "pobres_lpi"] = 1
lp_ingresos20.loc[(((lp_ingresos20["ict_percapita"] >= Lineas_pobreza[(2020,"Rural")]["Linea de Pobreza"]) & (lp_ingresos20["rural"] == 1)) & ~(lp_ingresos20["ict_percapita"].isna())) |
                  (((lp_ingresos20["ict_percapita"] >= Lineas_pobreza[(2020,"Urbana")]["Linea de Pobreza"]) & (lp_ingresos20["rural"] == 0)) & ~(lp_ingresos20["ict_percapita"].isna())), "pobres_lpi"] = 0
# 2022
lp_ingresos22.loc[((lp_ingresos22["ict_percapita"] < Lineas_pobreza[(2022,"Rural")]["Linea de Pobreza"]) & (lp_ingresos22["rural"] == 1)) |
                  ((lp_ingresos22["ict_percapita"] < Lineas_pobreza[(2022,"Urbana")]["Linea de Pobreza"]) & (lp_ingresos22["rural"] == 0)), "pobres_lpi"] = 1
lp_ingresos22.loc[(((lp_ingresos22["ict_percapita"] >= Lineas_pobreza[(2022,"Rural")]["Linea de Pobreza"]) & (lp_ingresos22["rural"] == 1)) & ~(lp_ingresos22["ict_percapita"].isna())) |
                  (((lp_ingresos22["ict_percapita"] >= Lineas_pobreza[(2022,"Urbana")]["Linea de Pobreza"]) & (lp_ingresos22["rural"] == 0)) & ~(lp_ingresos22["ict_percapita"].isna())), "pobres_lpi"] = 0

# Exportar bases
lp_ingresos16.to_csv("Linea de Pobreza por Ingresos 2016.csv", encoding="utf-8",index=False)
lp_ingresos18.to_csv("Linea de Pobreza por Ingresos 2018.csv", encoding="utf-8",index=False)
lp_ingresos20.to_csv("Linea de Pobreza por Ingresos 2020.csv", encoding="utf-8",index=False)
lp_ingresos22.to_csv("Linea de Pobreza por Ingresos 2022.csv", encoding="utf-8",index=False)
```
